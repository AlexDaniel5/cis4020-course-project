```python
DATA_SET_URL = "https://archive.ics.uci.edu/static/public/350/default+of+credit+card+clients.zip"

DATASET_FILE_ZIP = "default_of_credit_card_clients.zip"

DATASET_FILE = "default of credit card clients.xls"
```

```python
import requests
import pandas as pd
import zipfile
import numpy as np
```

```python
#using requests to download the dataset
r = requests.get(DATA_SET_URL)
with open(DATASET_FILE_ZIP, "wb") as f:
    f.write(r.content)

#unzipping the dataset
with zipfile.ZipFile(DATASET_FILE_ZIP, "r") as zip_ref:
    zip_ref.extractall()
    
#reading the dataset
df = pd.read_excel(DATASET_FILE, skiprows=1)
```

```python
#print the first 5 rows of the dataset
print("---First 5 Rows---")
print(df.head())

#print shape 
print("---Shape---")
print(df.shape)

#print summary of the dataset
print("---Summary---")
print(df.describe())

#print columns of the dataset
print("---Columns---")
print(df.columns)

#pirnt number of rows 
print("---Number of Rows---")
print(df.shape[0])
```
